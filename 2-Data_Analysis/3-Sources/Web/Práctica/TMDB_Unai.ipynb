1. Consultar el endpoint de "Genre List" para obtener la relación entre IDs y nombres.
2. Sustituir los IDs en tu DataFrame final por los nombres reales de los géneros (separados por comas).

````python
movies = search_movies(token, mi_inicial)["results"]
movies_U = [ movie for movie in movies if movie["title"].lower().startswith("u")]
````

````python
movies_df = pd.DataFrame(movies_U)
````

````python
def get_genre_str(genre_ids: str, genres: list):
    genre_names = []
    if len(genre_ids) > 0:
        for i in genre_ids:
            for genre in genres:
                if i == genre.get("id"):
                    genre_names.append(genre.get("name"))
            
    return ",".join(genre_names)
    
````

````python
genre_names_map: dict = {}

for genre_ids in movies_df["genre_ids"]:
    key = str(genre_ids).replace("[","").replace("]", "").replace(" ", "")
    genre_names_map[key] = get_genre_str(genre_ids, genres)

pos = movies_df.columns.get_loc("genre_ids") + 1
movies_df.insert(pos, "genre_names", movies_df["genre_ids"].astype(str).str.replace(r"[\[\]\s]", "", regex=True).map(genre_names_map))
#movies_df["genre_names"] = movies_df["genre_ids"].astype(str).str.replace(r"[\[\]\s]", "", regex=True).map(genre_names_map)
````
