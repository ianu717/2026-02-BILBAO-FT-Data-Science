```python
import sqlite3
import pandas as pd
```

```python
#Creamos la conexión y cursor
connection = sqlite3.connect("./cervezas_db.sqlite")
cursor = connection.cursor()
```

```python
#Creamos las tablas mediante el script
with open('./create_cervezas_schema.sql','r', encoding = 'utf-8') as s:
    create_schema_script = s.read()
cursor.executescript(create_schema_script)
```

```python
for i in range(1, 11):
    with open(f'./Ej{i}.sql','r', encoding = 'utf-8') as s:
        enunciado = s.readline().replace('-' , '').strip()
        script = s.read()
    print(f'\033[94m{enunciado}\033[0m')
    print(f'\033[96m{script}\033[0m')
    if i <= 8:
        cursor.execute(script)
        data = cursor.fetchall()
        col_names = [col_name[0] for col_name in cursor.description]
        print(f'\033[92mData Frame:\033[0m')
        print(f'\033[92m{pd.DataFrame(data, columns = col_names)}\033[0m')
    else:
        #para los ejercicios 9 y 10 usamos executescript() por que son update e insert
        cursor.executescript(script)
    print()
```

```python
connection.close()
```
