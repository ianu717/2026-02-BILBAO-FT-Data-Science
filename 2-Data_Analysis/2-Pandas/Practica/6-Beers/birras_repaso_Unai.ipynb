## Ejercicio 11
No queda muy claro cuánto de dispar, de dispersa, es la medida de IBU. Mediante estadística descriptiva (una o dos métricas, sin gráficas), razona la dispersión de la variable IBU.

```python
ibu_nulls = round((beer["ibu"].isna().sum() / len(beer)) * 100, 2)
ibu_nulls
```

```python
stats = beer["ibu"].agg(["mean", "std"])
coef_var = stats["std"] / stats["mean"] 

print(f"El coeficiente de variacion es del {round(coef_var * 100, 2)}%")
print(f"Podemos decir que la medida del IBU es muy dispar\naunque hay que tener en cuenta que la columna tiene un {ibu_nulls}% de valore nulos")
```

Consigue en un DataFrame de todas las cervezas en cuyas cervecerías al menos haya una cerveza con un `abv` mínimo 0.07.
