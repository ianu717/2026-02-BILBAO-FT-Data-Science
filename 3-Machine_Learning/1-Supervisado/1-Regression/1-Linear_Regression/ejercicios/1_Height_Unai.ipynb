### ¿Es una buena técnica para este tipo de problemas? Demuéstralo antes de entrenar un modelo con un análisis descriptivo.

```python
df = pd.DataFrame(data=lista_alumnos, columns=['nombre','edad','altura'])
df
```

- Las variables edad y altura estan debilmente correlacionadas. Son viables para alimentar a un modelo.

### Entrena un modelo de regresión lineal

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
x = df[['edad']]
y = df['altura']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.33, random_state=42)
```

```python
lm = LinearRegression()
lm.fit(x_train, y_train)
```

### ¿Qué error se comete? Calcula los errores que está cometiendo tu modelo de forma manual con python. Aparte, usa las métricas de sklearn MAE, MAPE, MSE y el RMSE.

```python
print(lm.intercept_)
```

```python
print(lm.coef_)
```

```python
coeff_df = pd.DataFrame(lm.coef_, x.columns, columns=['coefficient'])
coeff_df
```

```python
x_test
```

```python
predictions = lm.predict(x_test)
predictions
```

```python
sns.scatterplot(x = y_test, y = predictions)
```

```python
sns.displot((y_test - predictions));
```

```python
from sklearn import metrics
import numpy as np
print('MAE:', metrics.mean_absolute_error(y_test, predictions))
print('MSE:', metrics.mean_squared_error(y_test, predictions))
print('RMSE:', np.sqrt(metrics.mean_squared_error(y_test, predictions)))
print(metrics.mean_absolute_percentage_error(y_test, predictions))
```
